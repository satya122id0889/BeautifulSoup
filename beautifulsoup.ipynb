```python
import requests

url = 'https://en.wikipedia.org/wiki/Blog'
response = requests.get(url)
html = response.content
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html, 'lxml')  # 'html.parser' is slower, 'lxml' is faster
```

```python
title_tag = soup.find('title')
print(title_tag.text)
```

```python
all_links = soup.find_all('a')
for link in all_links:
    print(link.get('href'))
```

```python
div = soup.find('div', class_='my-class')
heading = soup.find('h1', id='main-title')
```

```python
print(div , heading)
```

```python
items = soup.select('div.container > ul > li')
for item in items:
    print(item.text)
```

```python
from bs4 import BeautifulSoup

import requests
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'
soup = BeautifulSoup(requests.get(url).text , 'html')
```

```python
table = soup.find_all('table')[0]
table
```

```python
table_titles = table.find_all('th')
table_titles
```

```python
table_titles = [table_title.text.strip() for table_title in table_titles]
table_titles
```

```python
import pandas as pd
df = pd.DataFrame(columns = table_titles)
df
```

```python
 column_datas = table.find_all('tr')[1:]
 column_datas
```

```python
for column_data in column_datas:
  row_data = column_data.find_all('td')
  row_data = [rd.text.strip() for rd in row_data]
  length = len(df)
  df.loc[length] = row_data
```

```python
df
```

```python
df.to_csv('companies_revenue_usa.csv', index=False)
```
